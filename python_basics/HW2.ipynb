```python
temperatures =  [22, 25, 19, 23, 26, 20, 21, 22]
```

```python
print(f"First el: {temperatures[0]}")
print(f"Last el: {temperatures[-1]}")
```

```python
temperatures.append(24)
temperatures.remove(19)
```

```python
print(f"Length of temperatures: {len(temperatures)}")
```

```python
summ = sum(temperatures)
max_temp = max(temperatures)
```

```python
print(f"Sum of temperatures: {summ}")
print(f"Maximum temperature: {max_temp}")
```
