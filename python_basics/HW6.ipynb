```python
def show_menu(dishes):
    print('=== MENU ===')
    for i, dish in enumerate(dishes, start=1):
        print(f'{i}. {dish}')
    print('============')
```

```python
show_menu(['Mashroom soup', 'Chicken Kyiv', 'Lemonade'])
```

```python
def print_square(n):
    print(f'Square of {n} = {n * n}')
```

```python
print_square(5)
print_square(10)
```

```python
def multiply(a, b):
    return a * b
```

```python
print(multiply(3, 4))
print(multiply(5, 6))
```

```python
def find_max(numbers):
    #return max(numbers)
    max_n = numbers[0]
    for n in numbers:
        if n > max_n:
            max_n = n
    return max_n
```

```python
print(find_max([15, 42, 8, 23, 56, 19]))
```

```python
def check_grade(grade):
    if (grade >= 60):
        return 'Pass'
    else:
        return 'Fail'
```

```python
print(check_grade(60))
print(check_grade(45))
```
