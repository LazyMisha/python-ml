**Завдання 3.** Для того ж набору даних в змінній `sum_purchases` обчисліть **медіану** кожним з зазначених методів:  

  3.1. без використання жодних бібліотек, користуючись тільки Python.

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
sorted_purchases = sorted(sum_purchases)
len_purchases = len(sum_purchases)

if len_purchases % 2 == 0:
    median_1 = (sorted_purchases[len_purchases // 2 - 1] + sorted_purchases[len_purchases // 2]) / 2
else:
    median_1 = sorted_purchases[len_purchases // 2]

median_2 = np.median(sum_purchases)

print("Median (manual):", median_1)
print("Median (numpy):", median_2)

# I will use median when the data contains outliers because it better represents a typical observation. I would also use the median when the distribution is skewed, example: 1, 1, 1, 1, 1000. In sum_purchases I see 2 outliers that's why mean here is higher than median.
# I will use mean when I want to estimate the central tendency that depends on all values including outliers. In the current case if I wanna calculate the revenue - I will use mean.
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
mean_sum_purchases = sum(sum_purchases) / len(sum_purchases)
mean_visits = sum(visits) / len(visits)

covarience = sum([(sum_purchases[i] - mean_sum_purchases) * (visits[i] - mean_visits) for i in range(len(visits))]) / (len(visits) - 1)

std_deviation_sum = math.sqrt(sum([(x - mean_sum_purchases)**2 for x in sum_purchases]) / (len(sum_purchases) - 1))
std_deviation_visits = math.sqrt(sum([(x - mean_visits)**2 for x in visits]) / (len(visits) - 1))

corelation_1 = covarience / (std_deviation_sum * std_deviation_visits)
corelation_2 = np.corrcoef(sum_purchases, visits)[0, 1]

print("Corelation (manual): ", corelation_1)
print("Corelation (numpy): ", corelation_2)

# I see relatively weak negative linear correlation between visits and sums. Also there are outliers on scatter plot.
# I think it makes sense to remove outliers and calculate carrelation again to check if outliers impacts on it.
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
threshold = mean_sum_purchases + (std_deviation_sum * 2)

purchases_no_outliers = []
visits_no_outliers = []

for purchase, visit in zip(sum_purchases, visits):
    if (purchase <= threshold):
        purchases_no_outliers.append(purchase)
        visits_no_outliers.append(visit)

correlation = np.corrcoef(purchases_no_outliers, visits_no_outliers)[0, 1]

print("Correlation without outliers: ", correlation)

plt.scatter(purchases_no_outliers, visits_no_outliers)
plt.title('Scatter Plot of Purchases vs. Visits. No outliers.')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');

# The correlation changed from -0.303 to -0.368. This means that direction remained negative after removing outliers and correlation became stronger.
```
