1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

```python
def f(x):
    return 10*x-5
```

```python
derivative_f = (f(10) - f(2)) / (10 - 2)
derivative_f
```

```python
# I selected random values in a range of 0 to 10 to calculate derivative
# derivative of `f` is 10
```

```python
xs = np.linspace(0, 10, 10)

plt.plot(xs, f(xs), 'o')
plt.xlabel("Study hours")
plt.ylabel("Predicted grade")
plt.show()
```

```python
# in this case each value equals to 10
derivative_values = [derivative_f] * len(xs)

plt.plot(xs, derivative_values, 'o')
plt.xlabel("Study hours")
plt.ylabel("Grade change")
plt.show()
```

```python
# when increase a number of hours by 1 it adds +10 to the predicted grade
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')
f = sp.sin(x)
df = sp.diff(f, x)
df
```

```python
sin_at = sp.lambdify(x, f, np)
derivative_at = sp.lambdify(x, df, np)

xs = np.linspace(0, 10, 100)

plt.plot(xs, sin_at(xs), label="sin(x)")
plt.plot(xs, derivative_at(xs), label="Derivative")

plt.legend()
plt.show()                        
```

```python
f = sp.ln(x)
df = sp.diff(f, x)
df
```

```python
ln_at = sp.lambdify(x, f, np)
derivative_at = sp.lambdify(x, df, np)

# x value couldn't start from 0 because we can not devidi by 0
xs = np.linspace(0.1, 10, 100)

plt.plot(xs, ln_at(xs), label="ln_at(x)")
plt.plot(xs, derivative_at(xs), label="Derivative")

plt.legend()
plt.show()       
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')
f = x**2 - 5*x +1
df = sp.diff(f, x)
df
```

```python
expr_at = sp.lambdify(x, f, np)
derivative_at = sp.lambdify(x, df, np)

xs = np.linspace(0, 100, 100)

plt.plot(xs, expr_at(xs), label="x**2 - 5*x +1")
plt.plot(xs, derivative_at(xs), label="Derivative")

plt.legend()
plt.show()                        
```
