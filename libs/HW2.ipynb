```python
more_than_20 = tips_df[tips_df['total_bill'] > 20]
female_tips = tips_df[tips_df['sex'] == 'Female']
dinner_tips = tips_df[tips_df['time'] == 'Dinner']
high_value_dinner_tips = tips_df[(tips_df['time'] == 'Dinner') & (tips_df['total_bill'] > 30)]

def print_filtered_dataframes(data):
    print(f'Length: {len(data)}. Top 5 rows:\n{data.head()}')
```

```python
print_filtered_dataframes(more_than_20)
```

```python
print_filtered_dataframes(female_tips)
```

```python
print_filtered_dataframes(dinner_tips)
```

```python
print_filtered_dataframes(high_value_dinner_tips)
```
