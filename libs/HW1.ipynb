```python
import numpy as np
```

```python
array = np.array([5, 15, 25, 35, 45])
zeros = np.zeros(7)
ones = np.ones(6)
array0 = np.arange(10, 20)

print("Array:", array)
print("Zeros:", zeros)
print("Ones:", ones)
print("Array0:", array0)
```

```python
array = np.array([12, 24, 36, 48, 60])
doubled = array * 2
added = array + 10
powered = array ** 2
square_rooted = np.sqrt(array)

print("Doubled:", doubled)
print("Added:", added)
print("Powered:", powered)
print("Square rooted:", square_rooted)
```

```python
data = np.array([18, 27, 33, 41, 52, 68, 79])
mean = np.mean(data)
minn = np.min(data)
maxx = np.max(data)
summ = np.sum(data)

print("Mean:", mean)
print("Min:", minn)
print("Max:", maxx)
print("Sum:", summ)
```

```python
arr = np.array([15, 22, 35, 44, 58, 67]);
first = arr[0]
last = arr[-1]
elInd1 = arr[1]
elInd3 = arr[3]
biggerThan40 = arr[arr > 40]

print("First:", first)
print("Last:", last)
print("At index 1:", elInd1)
print("At index 3:", elInd3)
print("Bigger than 40:", biggerThan40)
```

```python
matrix = np.array([
    [2, 4, 6], 
    [8, 10, 12],
    [14, 16, 18]])

shape = matrix.shape
firstRow = matrix[0, :]
secondCol = matrix[:, 1]
summ = np.sum(matrix)
mean = np.mean(matrix)

print("Matrix shape:", shape)
print("First row:", firstRow)
print("Second column:", secondCol)
print("Sum:", summ)
print("Mean:", mean)
```
